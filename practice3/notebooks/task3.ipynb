```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("../../data/country_stats-selected-columns.csv")
```

```python
df.shape
```

```python
categorical_numbers  = ["Country name"]

numerical_columns = [
    'Total in km2',
    'Land in km2',
    'Inland water in km2',
    '% water',
    'Population',
    '% of world population',
    'Population density',
    'HDI',
    'Fertility rate'
]

```

```python
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
```

```python
preprocessor = ColumnTransformer(
    transformers = [
        ('num', numeric_transformer, numerical_columns),
    ]
)
```

```python
x_transformed = preprocessor.fit_transform(df)
```

```python
x_transformed.shape
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test = train_test_split(
    df,
    test_size=0.2,
    random_state=42
)
```

```python
x_train_transformed = preprocessor.fit_transform(x_train)
```

```python
x_test_transformed = preprocessor.transform(x_test)
```

```python
print("До преобразования:")
print("Train:", x_train.shape)
print("Test:", x_test.shape)

print("\nПосле преобразования:")
print("Train:", x_train_transformed.shape)
print("Test:", x_test_transformed.shape)
```
